## 4. Benchmark Performance Visualization
Compare SpotifySupportAgent against the Simple and Trivial Baselines.

```python
import json
if os.path.exists('../eval/benchmark_metrics.json'):
    with open('../eval/benchmark_metrics.json') as f:
        metrics = json.load(f)
    
    summary = []
    for sys_name, data in metrics.items():
        summary.append({
            'System': sys_name,
            'Intent Accuracy (%)': data['intent']['accuracy'] * 100,
            'Intent Macro-F1 (%)': data['intent']['macro_f1'] * 100,
            'Escalation Recall (%)': data['escalation']['escalation_recall'] * 100
        })
    df_summary = pd.DataFrame(summary)
    display(df_summary)
```
